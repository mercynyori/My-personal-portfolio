```python
class Person:
  """example class: containing name and age"""
  def __init__(self, name, age):# self is always there
    self._name = name
    self._age = age
    pass
  def __str__(self):# tells a program how to write the output
    return f"name: {self._name}, age:{self._age}"# The underscore before the variable makes the class more private

```

```python
p1 = Person("Herman",19)
p1._age = 20
```

```python
person1 = Person("rafsan", 19)
person2 = Person("mercy", 18)
```

```python
class Circle:
  """ Class circle represents a circle"""
  total_circles = 0

  @classmethod
  def add_circle(cls):
    cls.total_circles += 1

  def __init__(self, radius = 1):
    self.radius = radius

    Circle.add_circle()

  def __str__(self):
    return f"circle: radius: {self.radius}, area:{self.area()}, total_circle: {Circle.total_circles}" #we call area with self.area

  def area(self):# methods always has self as the first parameters
    return self.radius * 2 * 3.1415




```

```python
c1 = Circle(2)
print(c1.radius)
c1.radius = 1
print(c1.radius)
```
